# Polynomial Regression

Linear Regression assumes that the relationship between input features
and target is linear.


The equation:

Price_lakhs =
β₀
+
β₁(Area_sqft)
+
β₂(Bedrooms)
+
β₃(Age_years)


represents a straight-line relationship.


However, in real-world problems, relationships may not always be linear.


For example:

- Increasing area may increase price faster after a certain point.
- The effect of bedrooms may depend on area.
- Age may have a nonlinear effect on price.


Polynomial Regression solves this by creating new features from the
existing features.


For degree = 2, new features include:


x₁²

x₂²

x₃²


and interaction terms:


x₁x₂

x₁x₃

x₂x₃



For our dataset:


x₁ = Area_sqft

x₂ = Bedrooms

x₃ = Age_years



The polynomial equation becomes:


Price_lakhs =

β₀

+ β₁Area

+ β₂Bedrooms

+ β₃Age

+ β₄Area²

+ β₅Bedrooms²

+ β₆Age²

+ β₇(Area × Bedrooms)

+ β₈(Area × Age)

+ β₉(Bedrooms × Age)


Polynomial Regression does not change the regression algorithm.

It changes the input features given to the model.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

df = pd.read_csv("house_regression_lab_dataset(in).csv")
x= df[["Area_sqft", "Bedrooms", "Age_years"]]
y = df["Price_lakhs"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

model = LinearRegression()

model.fit(
    x_train,
    y_train
)
y_pred = model.predict(x_test)
print("The predicted prices  are :",y_pred)
```

## Generating Polynomial Features


PolynomialFeatures from sklearn.preprocessing creates new features
by combining the original features.


Syntax:


PolynomialFeatures(degree=n)


where:


degree = maximum power of features.


For degree=2:


x → x²


and interaction terms are created.


The transformation:


X → X_poly


changes the feature representation.

```python
from sklearn.preprocessing import PolynomialFeatures


poly = PolynomialFeatures(
    degree=2
)

x_train_poly = poly.fit_transform(
    x_train
)

x_test_poly = poly.transform(
    x_test
)
```

## Display Feature Names


The transformed matrix contains numerical values, but we need names
to understand what each column represents.


get_feature_names_out()


returns the names of generated polynomial features.

```python
feature_names = poly.get_feature_names_out(
    [
        "Area_sqft",
        "Bedrooms",
        "Age_years"
    ]
)

feature_names
```

## Polynomial Regression Model


After feature transformation, Linear Regression is applied on the
new polynomial feature matrix.


The model learns:


y = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ


where each x represents a polynomial feature.


Each generated feature gets its own coefficient.

```python
from sklearn.linear_model import LinearRegression
poly_model = LinearRegression()
poly_model.fit(x_train_poly,y_train)
```

## Coefficient Calculation


The coefficients represent the contribution of each polynomial feature
towards predicting house price.


The order of coefficients follows:


feature_names


Therefore:


coefficient[0]

belongs to intercept term


remaining coefficients correspond to polynomial features.

```python
poly_coefficients = pd.DataFrame(
    {
        "Feature": feature_names,
        "Coefficient": poly_model.coef_
    }
)


poly_coefficients
```

## Polynomial Regression Intercept


The intercept represents the predicted value when all polynomial
features are zero.


The equation:


y = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ


β₀ is stored using:


model.intercept_

```python
print(poly_model.intercept_)
```

## Polynomial Regression Prediction


The trained polynomial model receives transformed test data:


X_test_poly


and calculates:


ŷ = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ


The predict() function generates predicted prices.

```python
y_poly_pred = poly_model.predict(
    x_test_poly)
y_poly_pred
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

import numpy as np


poly_mae = mean_absolute_error(
    y_test,
    y_poly_pred
)


poly_mse = mean_squared_error(
    y_test,
    y_poly_pred
)


poly_rmse = np.sqrt(
    poly_mse
)


poly_r2 = r2_score(
    y_test,
    y_poly_pred
)


poly_mae, poly_mse, poly_rmse, poly_r2
```

## Model Comparison


Two models are compared:


Linear Regression:

Polynomial Regression:

Comparison metrics:


RMSE:

Lower value is better.


R²:

Higher value is better.


If Polynomial Regression has lower RMSE and higher R²,
it performs better on test data.

```python
linear_results = pd.DataFrame(
    {
        "Model": [
            "Linear Regression",
            "Polynomial Regression"
        ],
        "RMSE": [
            rmse,
            poly_rmse
        ],
        "R2": [
            r2,
            poly_r2
        ]
    }
)


linear_results
```

## Effect of Polynomial Features


Polynomial features improve performance because they allow the model
to learn nonlinear relationships.


Example:


Linear Model:


Price = β₀ + β₁Area


Polynomial Model:


Price = β₀ + β₁Area + β₂Area²



The polynomial model can capture more complex patterns.


However, increasing polynomial degree creates many additional features.


Example:


Degree 2:

Area²


Degree 3:

Area³


Degree 4:

Area⁴



More features increase model complexity.


If the model learns noise instead of actual patterns,
overfitting occurs.


Overfitting means:


Training performance → Very high


Testing performance → Poor



Therefore:

Polynomial features can improve learning capability but excessive
complexity can reduce generalization.
