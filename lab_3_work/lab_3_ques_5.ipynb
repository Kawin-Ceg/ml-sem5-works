# Ridge Regression


Ridge Regression is a regularized version of Linear Regression.


Linear Regression tries to minimize only prediction error.


The cost function of Linear Regression is:


J(β) = Σ(y - ŷ)²



where:


y  → Actual value

ŷ  → Predicted value



Ridge Regression adds an additional penalty term to reduce large
coefficients.


The Ridge cost function is:


J(β) = Σ(y - ŷ)² + αΣβ²



The second term:


αΣβ²


is called L2 Regularization.


where:


α (alpha)

controls the strength of regularization.



β²

is the square of model coefficients.



Effect:


Small alpha:

Closer to Linear Regression


Large alpha:

More coefficient shrinkage



Ridge regression equation:


Price_lakhs =
β₀
+
β₁Area
+
β₂Bedrooms
+
β₃Age



but coefficients are calculated by minimizing:


Prediction Error + Coefficient Penalty

```python
from sklearn.linear_model import Ridge


ridge_model = Ridge(
    alpha=1
)


ridge_model.fit(
    x_train,
    y_train
)
```

## Ridge Intercept


The intercept represents the constant term:


y = β₀ + β₁x₁ + β₂x₂ + β₃x₃



β₀ is the predicted value when all features are zero.


The trained Ridge model stores this value using:


model.intercept_

```python
print(ridge_model.intercept_)
```

## Ridge Coefficients


The coefficients represent the contribution of each feature towards
predicting the house price.


The feature order is:


X = [

Area_sqft,

Bedrooms,

Age_years

]



Therefore:


ridge_model.coef_[0]

→ Area coefficient


ridge_model.coef_[1]

→ Bedroom coefficient


ridge_model.coef_[2]

→ Age coefficient



Ridge minimizes coefficients because of the L2 penalty:


αΣβ²

```python
print(ridge_model.coef_)
```

## Ridge Model Evaluation


After training, predictions are generated on unseen test data.


Prediction:


ŷ = Ridge Model(X_test)



Performance is evaluated using:


### MAE


MAE = (1/n)Σ|y - ŷ|



Measures average prediction error.



### MSE


MSE = (1/n)Σ(y - ŷ)²



Penalizes larger errors.



### RMSE


RMSE = √MSE



Same unit as house price.



### R²


R² = 1 - (SSres / SStot)



Measures how much variation is explained by the model.


```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

import numpy as np


ridge_pred = ridge_model.predict(
    x_test
)


ridge_mae = mean_absolute_error(
    y_test,
    ridge_pred
)


ridge_mse = mean_squared_error(
    y_test,
    ridge_pred
)


ridge_rmse = np.sqrt(
    ridge_mse
)


ridge_r2 = r2_score(
    y_test,
    ridge_pred
)


ridge_mae, ridge_mse, ridge_rmse, ridge_r2
```

# Why does Ridge Regression shrink coefficients?
## Coefficient Shrinkage in Ridge


Ridge adds a penalty:


αΣβ²



During optimization, large coefficients increase the penalty value.


Therefore, the model tries to reduce coefficient magnitude.


The objective becomes:


Minimize:


Error + Penalty



The model balances:


1. Accurate predictions

2. Smaller coefficients



Because the penalty uses squared coefficients:


β²


large values are strongly punished.


Example:


Without Ridge:


β = 100



Penalty:


100² = 10000



With Ridge:


β = 10



Penalty:


10² = 100



Therefore, Ridge prefers smaller coefficients.

# Effect of Ridge Regularization


Ridge Regression controls model complexity using the parameter alpha.


The Ridge objective function is:


J(β) = Σ(y - ŷ)² + αΣβ²


where:


α controls the strength of regularization.


Small alpha:

The penalty is weak.

The model behaves similar to Linear Regression.



Large alpha:

The penalty becomes stronger.

Coefficients shrink more.



The purpose is to study how changing alpha affects:


1. Feature coefficients

2. Prediction error

3. Model performance

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd


alpha_values = [0.01,0.1,1,10,100]


ridge_results = []


for alpha in alpha_values:

    ridge = Ridge(alpha=alpha)

    ridge.fit(
        x_train,
        y_train
    )

    pred = ridge.predict(
        x_test
    )

    rmse_value = np.sqrt(
        mean_squared_error(
            y_test,
            pred
        )
    )

    r2_value = r2_score(
        y_test,
        pred
    )

    ridge_results.append(
        [
            alpha,
            ridge.coef_[0],
            ridge.coef_[1],
            ridge.coef_[2],
            rmse_value,
            r2_value
        ]
    )
```

#### The results are stored in a table to observe the effect of alpha.

Columns:

Alpha:

Regularization strength


Area Coef:

Coefficient of Area_sqft


Bedrooms Coef:

Coefficient of Bedrooms


Age Coef:

Coefficient of Age_years


RMSE:

Prediction error


R²:

Model performance

```python
ridge_alpha_table = pd.DataFrame(
    ridge_results,
    columns=[
        "Alpha",
        "Area Coef",
        "Bedrooms Coef",
        "Age Coef",
        "RMSE",
        "R2"
    ]
)


print(ridge_alpha_table)
```

# Lasso Regression


Lasso Regression is another regularized regression technique.


Unlike Ridge which uses L2 penalty, Lasso uses L1 penalty.


Lasso cost function:


J(β)=Σ(y-ŷ)² + αΣ|β|



where:


|β|

represents absolute coefficient values.



Effect of Lasso:


- Reduces coefficient values
- Can make coefficients exactly zero
- Performs feature selection


Difference:


Ridge:

Shrinks coefficients


Lasso:

Shrinks and removes features

```python
from sklearn.linear_model import Lasso


lasso_model = Lasso(
    alpha=1
)


lasso_model.fit(
    x_train,
    y_train
)
```

```python
lasso_model.intercept_
lasso_model.coef_
lasso_pred = lasso_model.predict(
    x_test
)


lasso_mae = mean_absolute_error(
    y_test,
    lasso_pred
)


lasso_mse = mean_squared_error(
    y_test,
    lasso_pred
)


lasso_rmse = np.sqrt(
    lasso_mse
)


lasso_r2 = r2_score(
    y_test,
    lasso_pred
)


lasso_mae, lasso_mse, lasso_rmse, lasso_r2
```

# Ridge vs Lasso


## Ridge Regression

Penalty:

αΣβ²


Effect:

- Coefficients become smaller
- No feature elimination



## Lasso Regression

Penalty:

αΣ|β|


Effect:

- Coefficients shrink
- Some coefficients can become exactly zero


A zero coefficient means:


The feature does not contribute significantly to prediction.


Therefore Lasso performs automatic feature selection.

```python
lasso_results=[]


for alpha in alpha_values:

    lasso=Lasso(alpha=alpha)

    lasso.fit(
        x_train,
        y_train
    )

    pred=lasso.predict(
        x_test
    )


    rmse_value=np.sqrt(
        mean_squared_error(
            y_test,
            pred
        )
    )


    r2_value=r2_score(
        y_test,
        pred
    )


    lasso_results.append(
        [
            alpha,
            lasso.coef_[0],
            lasso.coef_[1],
            lasso.coef_[2],
            rmse_value,
            r2_value
        ]
    )
```

```python
lasso_table=pd.DataFrame(
    lasso_results,
    columns=[
        "Alpha",
        "Area Coef",
        "Bedrooms Coef",
        "Age Coef",
        "RMSE",
        "R2"
    ]
)


lasso_table
```

## 3. Coefficient Shrinkage


Linear Regression calculates coefficients by minimizing only prediction
error.


Ridge Regression adds an L2 regularization penalty:


J(β) = Σ(y - ŷ)² + αΣβ²


Because of this penalty, Ridge reduces the magnitude of coefficients.


Observation:

- Linear Regression has larger coefficients.
- Ridge Regression has smaller coefficients.


This helps reduce overfitting and improves model stability.

## 4. Feature Selection


Ridge and Lasso both use regularization, but their effects are different.


Ridge Regression:

- Uses L2 regularization.
- Shrinks coefficients.
- Does not make coefficients exactly zero.
- Keeps all features.


Lasso Regression:

- Uses L1 regularization.
- Can reduce some coefficients exactly to zero.
- Performs automatic feature selection.


A zero coefficient means that the corresponding feature does not
contribute significantly to the prediction.


This is useful when datasets contain many features because Lasso can
remove unnecessary features and create a simpler model.

## 5. Regularization Parameter


Alpha controls the strength of regularization.


### Ridge Regression


When alpha increases:

- The L2 penalty becomes stronger.
- Coefficients shrink more.
- Model complexity decreases.
- Very high alpha can cause underfitting.



### Lasso Regression


When alpha increases:

- The L1 penalty becomes stronger.
- More coefficients move towards zero.
- Some features may be completely removed.


Therefore:

Small alpha → Less regularization

Large alpha → Stronger regularization

## 6. Conceptual Comparison


| Model | Cost Function | Main Effect |
|---|---|---|
| Linear Regression | Σ(y - ŷ)² | Minimizes prediction error |
| Ridge Regression | Σ(y - ŷ)² + αΣβ² | Shrinks coefficients |
| Lasso Regression | Σ(y - ŷ)² + αΣ|β| | Performs feature selection |


Polynomial Regression changes the input features by creating new
features such as:

- Area²
- Bedrooms²
- Age²
- Area × Bedrooms


Ridge and Lasso do not change the features.

They modify the objective function used to estimate coefficients.
