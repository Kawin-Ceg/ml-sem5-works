# Multiple Linear Regression

Multiple Linear Regression is an extension of Linear Regression where
multiple independent variables are used to predict a single target variable.


In this problem:

Independent Variables:

X = [
    Area_sqft,
    Bedrooms,
    Age_years
]


Target Variable:

y = Price_lakhs


The model learns the relationship between input features and house price.


The mathematical representation of Multiple Linear Regression is:


y = β₀ + β₁x₁ + β₂x₂ + β₃x₃


where:


y  → Predicted house price


β₀ → Intercept

The value of y when all input features are zero.


β₁ → Coefficient of Area_sqft

Represents the change in price when area increases by one unit,
keeping other features constant.


β₂ → Coefficient of Bedrooms

Represents the change in price when the number of bedrooms increases
by one unit, keeping other features constant.


β₃ → Coefficient of Age_years

Represents the change in price when house age increases by one year,
keeping other features constant.


For this dataset:


Price_lakhs =
β₀
+
β₁(Area_sqft)
+
β₂(Bedrooms)
+
β₃(Age_years)


The model finds the best values of β₀, β₁, β₂ and β₃ by minimizing
the difference between actual and predicted prices.


The error function used is:


MSE = (1/n) Σ(y - ŷ)²


where:

y  → Actual value

ŷ  → Predicted value

n  → Number of samples

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
df = pd.read_csv("house_regression_lab_dataset(in).csv")
x= df[["Area_sqft", "Bedrooms", "Age_years"]]
y = df["Price_lakhs"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

model = LinearRegression()

model.fit(
    x_train,
    y_train
)
```

## Explanation of Model Training

LinearRegression() creates an empty regression model.

The fit() function trains the model using:

X_train → Input features

y_train → Actual house prices


During training, the model calculates:

β₀
β₁
β₂
β₃


which minimize the prediction error.


The learning process is:

Training Features
        |
        ↓
Linear Regression Algorithm
        |
        ↓
Find optimal coefficients
        |
        ↓
Trained Model

## Intercept Calculation

The intercept represents the base value of the prediction when all
features are zero.


The equation:


Price_lakhs =
β₀
+
β₁(Area_sqft)
+
β₂(Bedrooms)
+
β₃(Age_years)


Here:

β₀ = Intercept


The intercept is stored inside the trained model using:


model.intercept_

```python
print(model.intercept_)
```

## Regression Coefficients

Coefficients represent how much the target variable changes when
one feature changes by one unit while other features remain constant.


The coefficients are:


β₁ → Area_sqft coefficient

β₂ → Bedrooms coefficient

β₃ → Age_years coefficient


The model stores coefficients using:


model.coef_


The order of coefficients follows the order of columns in X:


X = [

Area_sqft,

Bedrooms,

Age_years

]


Therefore:


model.coef_[0] = β₁

model.coef_[1] = β₂

model.coef_[2] = β₃

```python
print(model.coef_)
```

## Final Regression Equation

Multiple Linear Regression follows the equation:


y = β₀ + β₁x₁ + β₂x₂ + β₃x₃


For this problem:


Price_lakhs =
β₀
+
β₁(Area_sqft)
+
β₂(Bedrooms)
+
β₃(Age_years)


The values of β₀, β₁, β₂, and β₃ are obtained from the trained model.


The final equation represents the relationship learned by the model.

It can be used to calculate the predicted house price for any new
combination of:

- Area
- Number of bedrooms
- Age of house

```python
print(
    f"Price_lakhs = {model.intercept_} + "
    f"({model.coef_[0]} * Area_sqft) + "
    f"({model.coef_[1]} * Bedrooms) + "
    f"({model.coef_[2]} * Age_years)"
)
```

## Interpretation

The printed equation is the mathematical model learned by the
Linear Regression algorithm.

Example:

Price_lakhs =
1.7
+
0.046(Area_sqft)
+
-0.5(Bedrooms)
-
0.08(Age_years)


This means:

- Area increases price
- More bedrooms increase price
- Older houses decrease price

according to the learned coefficients.

## Prediction Using Trained Model

After training, the model can predict prices for unseen data.


The prediction equation is:


ŷ = β₀ + β₁x₁ + β₂x₂ + β₃x₃


where:


ŷ → Predicted Price

β → Learned coefficients

x → Input features


The test dataset:

X_test

contains houses that the model has not seen during training.


The predict() function applies the learned equation and generates
predicted prices.

```python
y_pred = model.predict(x_test)
print("The predicted prices  are :",y_pred)
```

## Regression Evaluation Metrics

A regression model is evaluated by comparing:

Actual Values (y_test)

with

Predicted Values (y_pred)


### 1. Mean Absolute Error (MAE)


MAE = (1/n) Σ |y - ŷ|



MAE represents the average absolute difference between actual and
predicted values.


Lower MAE means better predictions.



### 2. Mean Squared Error (MSE)


MSE = (1/n) Σ(y - ŷ)²



MSE squares the errors before averaging.

Large errors are penalized more.



### 3. Root Mean Squared Error (RMSE)


RMSE = √MSE



RMSE is in the same unit as the target variable.

For this problem:

RMSE is in lakhs.



### 4. R² Score


R² = 1 - (Residual Sum of Squares / Total Sum of Squares)



R² measures how much variation in the target variable is explained
by the model.


Range:

1 → Perfect prediction

0 → Model explains no variation

Negative → Model performs worse than baseline

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

import numpy as np


mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)


print(mae)
print(mse)
print(rmse)
print(r2)
```
