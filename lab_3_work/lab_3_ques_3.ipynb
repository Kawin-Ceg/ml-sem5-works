## Splitting Dataset into Training and Testing Data

Machine learning models should not be evaluated on the same data
used for training.

Therefore, we split the dataset into:

Training Data:
80%

Used to learn patterns.

Testing Data:
20%

Used to evaluate model performance.

Parameters:
- test_size = 0.2
- random_state = 42

### Parameters

X:
Input features

y:
Target values

test_size=0.2:

20% data goes to testing.

Remaining 80% goes to training.


random_state=42:

Controls randomness.

Using the same value gives the same split every time.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
df = pd.read_csv("house_regression_lab_dataset(in).csv")
x= df[["Area_sqft", "Bedrooms", "Age_years"]]
y = df["Price_lakhs"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)


print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```
