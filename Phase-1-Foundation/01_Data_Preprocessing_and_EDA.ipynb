# Data Preprocessing and Exploratory Data Analysis (EDA)

## Understanding Data

A dataset consists of:

- Rows → individual observations
- Columns → features/attributes
- Target → output we want to predict
  

Data preprocessing is the process of converting raw,
incomplete and inconsistent data into a clean format
that machine learning algorithms can understand.


# Machine Learning Workflow Pipeline

The complete machine learning process follows these major stages:

```
Raw Data
    ↓
Data Understanding
    ↓
Exploratory Data Analysis (EDA)
    ↓
Data Cleaning
    ↓
Feature Engineering
    ↓
Data Transformation
    ↓
Machine Learning Model
```

## Workflow Explanation

1. **Raw Data**

   * Collection of original data from different sources.
   * Data may contain missing values, duplicates, inconsistencies, and noise.

2. **Data Understanding**

   * Understand the structure and characteristics of the dataset.
   * Analyze:

     * Number of rows and columns
     * Data types
     * Features and target variables
     * Basic statistics

3. **Exploratory Data Analysis (EDA)**

   * Analyze patterns, relationships, and distributions in data.
   * Includes:

     * Summary statistics
     * Data visualization
     * Correlation analysis
     * Outlier detection

4. **Data Cleaning**

   * Improve data quality by handling:

     * Missing values
     * Duplicate records
     * Incorrect data types
     * Outliers and inconsistencies

5. **Feature Engineering**

   * Create new meaningful features from existing data.
   * Select important features that improve model performance.

6. **Data Transformation**

   * Convert data into a suitable format for machine learning algorithms.
   * Includes:

     * Scaling
     * Encoding categorical variables
     * Normalization
     * Train-test splitting

7. **Machine Learning Model**

   * Train and evaluate machine learning algorithms.
   * Steps include:

     * Model selection
     * Training
     * Validation
     * Performance evaluation
     * Deployment

## Interview Questions

Q1. Why is preprocessing required before ML?

Answer:
Machine learning algorithms require structured numerical data.
Real-world data contains missing values, inconsistencies,
and noise which must be handled before training.

---

Q2. What is EDA?

Answer:
EDA is the process of analyzing data using statistics and
visualization to understand patterns, relationships and
data quality issues.

---

Q3. Difference between EDA and preprocessing?

Answer:

EDA:
Understanding and discovering patterns.

Preprocessing:
Modifying data into a model-ready format.
