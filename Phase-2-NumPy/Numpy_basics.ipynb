# Q: Why is NumPy faster than Python lists?

 ### Answer:

NumPy arrays store elements of the same datatype in contiguous memory locations and perform operations using optimized C implementations. This allows vectorized computation, making numerical operations much faster than Python loops over lists.

## np.arange()

Creates a NumPy array containing evenly spaced values.

### Syntax

```python
np.arange(start, stop, step)
```

The `stop` value is excluded.

Example:

```python
np.arange(0,10,2)
# [0 2 4 6 8]
```

## np.linspace()

Creates a fixed number of evenly spaced values between two endpoints.

### Syntax

```python
np.linspace(start, stop, num)
```

Example:

```python
np.linspace(0,1,5)
# [0.   0.25 0.5  0.75 1.  ]
```

Unlike `arange()`, the stop value is included by default.

# NumPy Indexing

## Why do we need indexing?

Indexing allows us to access specific elements from a NumPy array.

## Syntax

array[index]

NumPy uses zero-based indexing.

Example:

transactions[0]

returns the first element.

## Negative Indexing

Negative indexes access elements from the end.

Example:

transactions[-1]

returns the last element.

## Usage

Used for selecting specific observations,
extracting values, and analyzing subsets of data.


````python
import numpy as np
amounts = np.array([100,200,300,400,500])
print(amounts[1:4]) # array slicing is start:stop:step stop value is excluded..
print(amounts[:3])  # it defaultly means we start from 0 and end at 2 
print(amounts[1:]) # it tells the starting index but final stop value is not mentioned so it goes till the very last
print(amounts[::2]) # an example to just have the step value
````

# NumPy Slicing

## Why do we need it?

Slicing allows selecting a range or subset of elements
from a NumPy array.

## Syntax

array[start:stop:step]

start → beginning index
stop → ending index (excluded)
step → interval between elements

## Example

arr[1:4]

returns elements from index 1 to 3.

## Usage

Used for selecting portions of datasets,
sampling data, and analyzing subsets.

````python
import numpy as np

a = np.array([10,20,30,40,50,60])

print(a[2:5])
print(a[:4])
print(a[::3])
````

# Aggregate functions in numpy:



| Function | Purpose |
|----------|----------|

# Broadcasting

## Why?

Broadcasting allows NumPy to perform arithmetic operations on arrays of different but compatible shapes without writing loops.

## Example

```python
a = np.array([10,20,30])

a + 5
```

Output:

```text
[15 25 35]
```

NumPy conceptually broadcasts `5` to `[5,5,5]`.

## Usage

- Data normalization
- Scaling
- Feature transformations
- Machine learning preprocessing

## Important Notes

- Broadcasting is automatic.
- It requires compatible shapes.
- It improves performance by avoiding explicit Python loops.

````python
import numpy as np
a = np.array([5,10,15])
b = np.array([2,4,6])
print(a + b)
print(a * 3)
````

````python
import numpy as np
a = np.array([
    [1,2,3],
    [4,5,6]
])
b = np.array([10,20,30])
print(a + b)
````
